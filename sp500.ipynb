```python
import pandas as pd
```

```python
df = pd.read_csv('sp500.csv')
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Rok'] = df['Date'].dt.year
```

```python
df['Miesiąc'] = df['Date'].dt.month
```

```python
df['Zmiana_Procentowa'] = df.groupby('Ticker')['Adj_Close'].pct_change()
```

```python
df['Trend'] = df['Zmiana_Procentowa'].apply(lambda x: 'Wzrost' if x > 0 else ('Spadek' if x <= 0 else 'Brak danych'))
```

```python
df.to_csv('sp500_przetworzone.csv', index=False)
```

```python
tabela_roczna = df.groupby(['Ticker', 'Rok'])['Adj_Close'].mean().reset_index()
```

```python
tabela_roczna.to_csv('sp500_sr_roczne.csv', index=False)
```

```python
logo_urls = {
    'AAPL': 'https://upload.wikimedia.org/wikipedia/commons/f/fa/Apple_logo_black.svg',
    'AMZN': 'https://upload.wikimedia.org/wikipedia/commons/a/a9/Amazon_logo.svg',
    'AVGO': 'https://upload.wikimedia.org/wikipedia/en/thumb/f/f4/Broadcom_Inc._Logo.svg/1200px-Broadcom_Inc._Logo.svg.png',
    'GOOG': 'https://upload.wikimedia.org/wikipedia/commons/2/2f/Google_2015_logo.svg',
    'GOOGL': 'https://upload.wikimedia.org/wikipedia/commons/2/2f/Google_2015_logo.svg',
    'MSFT': 'https://upload.wikimedia.org/wikipedia/commons/4/44/Microsoft_logo.svg',
    'NVDA': 'https://upload.wikimedia.org/wikipedia/commons/a/a4/NVIDIA_logo.svg',
    'TSLA': 'https://upload.wikimedia.org/wikipedia/commons/e/e8/Tesla_logo.png',
    'META': 'https://upload.wikimedia.org/wikipedia/commons/7/7b/Meta_Platforms_Inc._logo.svg'
}
```

```python
df['Logo_URL'] = df['Ticker'].map(logo_urls)
```

```python
df.to_csv('sp500_przetworzone.csv', index=False)
```

```python
df.loc[df['Ticker'] == 'AVGO', 'Logo_URL'] = 'https://upload.wikimedia.org/wikipedia/commons/4/47/Broadcom_logo_%281991-2016%29.svg'
```

```python
df.to_csv('sp500_przetworzone.csv', index=False)
```

```python
df['Cena_Startowa'] = df.groupby('Ticker')['Adj_Close'].transform('first')
```

```python
df['Wynik_Indexowany'] = (df['Adj_Close'] / df['Cena_Startowa'] * 100)
```

```python
df.to_csv('sp500_przetworzone.csv', index=False)
```

```python
stats = df.groupby('Ticker')['Adj_Close'].agg(['first', 'last'])
```

```python
stats['Calkowity_Zwrot'] = (stats['last'] / stats['first'] - 1)
```

```python
daily_returns = df.groupby('Ticker')['Adj_Close'].pct_change()
daily_returns_clean = daily_returns[daily_returns.abs() < 0.5]
risk = daily_returns_clean.groupby(df['Ticker']).std() * (252**0.5)
```

```python
kpi = pd.concat([stats, risk], axis=1).reset_index()
```

```python
kpi.columns = ['Ticker', 'Cena_Start', 'Cena_Koniec', 'Calkowity_Zwrot', 'Ryzyko_Zmiennosc']
```

```python
kpi['Logo_URL'] = kpi['Ticker'].map(logo_urls)
```

```python
kpi.loc[kpi['Ticker'] == 'AVGO', 'Logo_URL'] = 'https://upload.wikimedia.org/wikipedia/commons/4/47/Broadcom_logo_%281991-2016%29.svg'
```

```python
kpi.to_csv('sp500_kpi_risk.csv', index=False)
```

```python
google_new = 'https://upload.wikimedia.org/wikipedia/commons/thumb/c/c1/Google_%22G%22_logo.svg/512px-Google_%22G%22_logo.svg.png'
```

```python
df.loc[df['Ticker'].isin(['GOOG', 'GOOGL']), 'Logo_URL'] = google_new
```

```python
kpi.loc[kpi['Ticker'].isin(['GOOG', 'GOOGL']), 'Logo_URL'] = google_new
```

```python
df.to_csv('sp500_przetworzone.csv', index=False)
```

```python
kpi.to_csv('sp500_kpi_risk.csv', index=False)
```

```python
annual_return = daily_returns_clean.groupby(df['Ticker']).mean() * 252
```

```python
risk_free_rate = 0.0425
```

```python
sharpe = (annual_return - risk_free_rate) / risk
```

```python
kpi['Sharpe_Ratio'] = kpi['Ticker'].map(sharpe)
```

```python
kpi.to_csv('sp500_kpi_risk.csv', index=False)
```
